# Dictionaries

```python
student ={'name':'swathi','age':18,'course':['computerscience','maths']}
print(student)
```

```python
print(student['course'])
#get() is most commonly used with dictionaries to safely retrieve the value associated with a key
print(student.get('name'))
student['phone'] ='555-566'
print(student)
```

```python
student.update({'name':'nitya','age':15})
print(student)
```

```python
#to add element in dict we use.update()
#to delet the elemnt we use del() and pop()
del student['phone']
print(student)
```

```python
student.pop('age')
print(student)
```

```python
#retrival in dict
print(len(student))
print(student.keys())
print(student.values())
print(student.items())
```

```python
# using loop in dict
for key ,value in student.items():
  print(key,value)
```

```python
users = {
    "admin": "1234",
    "swathi": "abcd",
    "student": "pass123"
}

username = input("Enter Username: ")
password = input("Enter Password: ")

if username in users:
    if users[username] == password:
        print("Login Successful!")
    else:
        print("Incorrect Password!")
else:
    print("Username Not Found!")
```

```python
users = {
    "admin": "1234",
    "swathi": "abcd"
}

while True:
    print("\n===== LOGIN SYSTEM =====")
    print("1. Login")
    print("2. Sign Up")
    print("3. Change Password")
    print("4. Delete Account")
    print("5. Display Users")
    print("6. Exit")

    choice = input("Enter your choice: ")

    # Login
    if choice == "1":
        attempts = 3

        while attempts > 0:
            username = input("Enter Username: ")
            password = input("Enter Password: ")

            if username in users and users[username] == password:
                print("Login Successful!")
                break
            else:
                attempts -= 1
                print("Invalid Username or Password")
                print("Attempts Left:", attempts)

        if attempts == 0:
            print("Account Locked! Try Again Later.")

    # Sign Up
    elif choice == "2":
        username = input("Enter New Username: ")

        if username in users:
            print("Username Already Exists!")
        else:
            password = input("Enter Password: ")
            users[username] = password
            print("Account Created Successfully!")

    # Change Password
    elif choice == "3":
        username = input("Enter Username: ")

        if username in users:
            old_password = input("Enter Old Password: ")

            if users[username] == old_password:
                new_password = input("Enter New Password: ")
                users[username] = new_password
                print("Password Changed Successfully!")
            else:
                print("Incorrect Old Password!")
        else:
            print("Username Not Found!")

    # Delete Account
    elif choice == "4":
        username = input("Enter Username: ")

        if username in users:
            password = input("Enter Password: ")

            if users[username] == password:
                del users[username]
                print("Account Deleted Successfully!")
            else:
                print("Incorrect Password!")
        else:
            print("Username Not Found!")

    # Display Users
    elif choice == "5":
        print("\nRegistered Users:")
        for username in users.keys():
            print(username)

    # Exit
    elif choice == "6":
        print("Thank You!")
        break

    else:
        print("Invalid Choice!")
```
