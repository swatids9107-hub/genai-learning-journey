```python
# lower(0) method coverts the string into lower case
swan = "Swan is Loyal bird"
print(swan.lower())
```

```python
# upper() method converts the string into upper case
swan = "swan is loyal bird"
print(swan.upper())
```

```python
#Count() method counts the number of occurences of a substring in a string
swan = "swan is loyal bird"
print(swan.count('bird'))
print(swan.count('l'))
```

```python
# we can find the index of where certain characters can be found and we can use the find method to find the index of a substring  in a string
swan = " swam is loyal bird"
print(swan.find('bird'))
print(swan.find("swan"))# output -1 bcoz swan is not prsent ith the string
```

```python
#replace() method replaces a substring with another  substring in a main string
swan = "swan is loyal bird"
print(swan.replace("loyal","faithful"))
```

```python
# addind two string useing + symbol
greeting = 'Hello'
name = 'Swan'

message = greeting + ', '+ name + '. Welcome swan !'
print(message)
```

```python
# format() method is used to format the string and it is using the {} to replace the valuse 
greeting = 'Hello'
name = 'swan'
message = '{}, {}. Welocme swan!'.format(greeting, name)
print(message)
```

```python
# fstrings same as the format() method placeholder
message = f'{greeting},{name.upper()}.welcome!'
print(message)
```
